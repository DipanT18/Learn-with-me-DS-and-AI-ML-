```python
#Importing dependencies
import gymnasium
import flappy_bird_gymnasium
import pygame
import warnings
warnings.filterwarnings("ignore", category=UserWarning)

pygame.init()
```

```python
#Create a environment in human rendering mode
try:
    env = gymnasium.make("FlappyBird-v0", render_mode = "human")
except pygame.error as e:
    print(f"Pygame initialization failed: {e}")
    env.close()
    exit()

#Reset the environment to start a new game and get the initial state observation
obs, info = env.reset()
```

```python
#Loop until the game is over
while True:
    #The agent selects a random action (0 = do nothing, 1= flap)
    action = env.action_space.sample()
    # Execute the action and get the next state, reward, termination status, and info
    obs, reward, terminated, truncated, info = env.step(action)
    # The game ends if the bird hits a pipe, the ground, or the top of the screen
    if terminated or truncated:
        print("Game Over!")
        break

# Close the environment
env.close() 
pygame.quit()  
```
