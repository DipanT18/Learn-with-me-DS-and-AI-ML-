```python
import numpy as np
import time
import pandas as pd
```

```python
'''Here X1 =  size of the house in sqft, X2 = number of bedrooms, X3 = age of the house in years, X4 = distance from city center in miles, X5 = number of floors, Y = price of the house in $'''
df = pd.DataFrame({
    "X1": [2050, 1600, 2400, 1416, 3000, 1985, 1534, 1427, 1380, 1494],
    "X2": [3, 3, 3, 2, 4, 3, 3, 2, 2, 3],
    "X3": [10, 15, 5, 20, 8, 12, 18, 25, 30, 14],
    "X4": [5, 10, 3, 15, 7, 6, 12, 20, 25, 8],
    "X5": [2, 1, 2, 1, 3, 2, 1, 1, 1, 2],
    "Y": [500000, 350000, 600000, 250000, 750000, 450000, 300000, 200000, 180000, 400000]
})
```

```python
#Split the data into features and target variable
X = df[["X1", "X2", "X3", "X4", "X5"]].values
Y = df["Y"].values
```

```python
#Splitting data into training and testing sets
#But we are not using sckit-learn, so we will do it manually
# train_size = int(0.8 * len(X))
# X_train, X_test = X[:train_size], X[train_size:]
# Y_train, Y_test = Y[:train_size], Y[train_size:]
```

```python
m = len(Y)  # number of training examples
n = X.shape[1]  # number of features
X = np.hstack((np.ones((m, 1)), X))  # Add intercept term
theta = np.zeros(n + 1)  # Initialize parameters
```

```python
#Now defining models manually 
def predict(X, theta):
    return X.dot(theta)
    
def compute_cost(X, Y, theta):
    m = len(Y)
    predictions = predict(X, theta)
    errors = predictions - Y
    cost = (1 / (2 * m)) * np.sum(errors ** 2)
    return cost
def gradient_descent(X, Y, theta, alpha, iterations):
    m = len(Y)
    cost_history = []

    for i in range(iterations):
        predictions = predict(X, theta)
        errors = predictions - Y
        gradients = (1 / m) * X.T.dot(errors)
        theta -= alpha * gradients
        cost = compute_cost(X, Y, theta)
        cost_history.append(cost)

    return theta, cost_history
```

```python
initial_theta = np.zeros((n + 1, 1))
alpha = 0.0000001  # Learning rate
iterations = 1000  # Number of iterations
print(f"Initial Cost: {compute_cost(X, Y, initial_theta):.2f}")
```

```python
#Fitting the model
optimal_theta, cost_history = gradient_descent(X, Y, theta, alpha, iterations)
print(f"Optimal Parameters: {optimal_theta}")
print(f"Final Cost: {cost_history[-1]:.2f}")
#Making predictions
predictions = predict(X, optimal_theta)
print("Predictions:", predictions)
```
